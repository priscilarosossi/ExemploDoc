Observamos que algumas colunas distoam entre a média e a mediana, principalmente, colunas de compras. Comportamento clássico em dados de consumo real: alguns poucos clientes gastam muito mais que a maioria, o que puxa a média (mean) para cima. Nestes casos, **a distribuição é assimétrica à direita (right-skewed)**.

Para o marketing isso pode indicar clientes VIPs ou heavy users — **que podem ser interessantes para campanhas específicas.**

Iremos usar a **mediana** para entender o comportamento da maioria dos clientes e a **média** se o objetivo for entender o faturamento total (pois esses grandes compradores influenciam mais no lucro).

*Podemos crie segmentos:*

*Um grupo “regular” (clientes comuns, gastos até o 75º percentil);*

*Um grupo “premium” (clientes acima do 75º ou 90º percentil).*

```python
p50 = df['MntTotal'].quantile(0.50)
p75 = df['MntTotal'].quantile(0.75)

# Criar função para classificar os clientes
def classificar_cliente(valor):
    if valor <= p50:
        return 'Baixo gasto'
    elif valor <= p75:
        return 'Gasto médio'
    else:
        return 'Alto gasto'

# Aplicar a função e criar nova coluna
df['segmento_cliente'] = df['MntTotal'].apply(classificar_cliente)

# Verificar a distribuição dos segmentos
df['segmento_cliente'].value_counts(normalize=True) * 100
```

* Podemos observar que metade dos clientes faz poucas compras ou gasta pouco pr mês.
* Um quarto dos clientes tem consumo moderado, comprando com frequência ou valores medianos.
* Quase um quarto é formado por clientes premium, que gastam bastante.

AÇÃO DE MARKETING

**A maior parte da base (50%) é de baixo gasto.**
→ Estratégias de marketing podem focar em aumentar o ticket médio desse grupo com promoções, cupons ou combos.

**Os 25% de alto gasto merecem atenção especial.**
→ Esses clientes são fiéis e valiosos — vale personalizar campanhas, oferecer benefícios exclusivos ou planos de fidelidade.

**Há uma boa base intermediária (25%).**
